```python
import pandas as pd

df = pd.read_excel(
    "../data/Kittchen PNL Data.xlsx",
    header=1
)

df.columns = (
    df.columns
    .str.strip()
    .str.upper()
    .str.replace(" ", "_")
)

df.head()
```

```python
# Gross Margin %
df["GM_PERCENT"] = (
    df["GROSS_MARGIN"] / df["NET_REVENUE"]
) * 100

# EBITDA %
df["EBITDA_PERCENT"] = (
    df["KITCHEN_EBITDA"] / df["NET_REVENUE"]
) * 100

# Variance %
df["VARIANCE_PERCENT"] = (
    df["VARIANCE"] / df["NET_REVENUE"]
) * 100

df.head()
```

```python
total_revenue = df["NET_REVENUE"].sum()

total_gross_margin = df["GROSS_MARGIN"].sum()

total_ebitda = df["KITCHEN_EBITDA"].sum()

total_orders = df["ORDER_COUNT"].sum()

print("Total Revenue :", total_revenue)

print("Total Gross Margin :", total_gross_margin)

print("Total EBITDA :", total_ebitda)

print("Total Orders :", total_orders)
```

```python
city_summary = df.groupby("CITY")[
    ["NET_REVENUE", "GROSS_MARGIN", "KITCHEN_EBITDA"]
].sum().reset_index()

city_summary
```

```python
import plotly.express as px

fig = px.bar(
    city_summary,
    x="CITY",
    y="NET_REVENUE",
    color="CITY",
    title="Revenue by City"
)

fig.show()
```

```python
fig = px.pie(
    city_summary,
    names="CITY",
    values="KITCHEN_EBITDA",
    title="EBITDA Contribution by City"
)

fig.show()
```

```python
top_stores = df.groupby("STORE")[
    "NET_REVENUE"
].sum().reset_index()

top_stores = top_stores.sort_values(
    by="NET_REVENUE",
    ascending=False
)

top_stores.head(10)
```

```python
fig = px.bar(
    top_stores.head(10),
    x="STORE",
    y="NET_REVENUE",
    color="NET_REVENUE",
    title="Top 10 Revenue Stores"
)

fig.show()
```

```python
variance_summary = df.groupby(
    "REVENUE_COHORT"
)["VARIANCE_PERCENT"].mean().reset_index()

variance_summary
```

```python
fig = px.bar(
    variance_summary,
    x="REVENUE_COHORT",
    y="VARIANCE_PERCENT",
    color="REVENUE_COHORT",
    title="Average Variance % by Revenue Cohort"
)

fig.show()
```

```python
fig = px.scatter(
    df,
    x="NET_REVENUE",
    y="KITCHEN_EBITDA",
    color="CITY",
    size="ORDER_COUNT",
    hover_name="STORE",
    title="Revenue vs EBITDA"
)

fig.show()
```

```python
def variance_bucket(x):

    if x < 2:
        return "Below 2%"

    elif x < 3:
        return "2% to 3%"

    elif x < 5:
        return "3% to 5%"

    else:
        return "Above 5%"


df["VARIANCE_CATEGORY"] = df[
    "VARIANCE_PERCENT"
].apply(variance_bucket)

df.head()
```

```python
pivot_table = pd.pivot_table(
    df,
    values="STORE",
    index="VARIANCE_CATEGORY",
    columns="MONTH",
    aggfunc="count",
    fill_value=0
)

pivot_table
```
